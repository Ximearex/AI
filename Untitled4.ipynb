Para la obtención de las métricas que se presentan a continuación, se implementó el código desarrollado en Python. El proceso partió de la base matemática establecida mediante la función trigonométrica principal (*1. Definición de la función para calcular el ángulo*).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Definición de la función para calcular el ángulo
def angle_deg(v1, v2):
    v1 = np.asarray(v1, dtype=float)
    v2 = np.asarray(v2, dtype=float)
    dot = np.sum(v1 * v2, axis=-1)
    n1 = np.linalg.norm(v1, axis=-1)
    n2 = np.linalg.norm(v2, axis=-1)
    cosang = dot / (n1 * n2)
    cosang = np.clip(cosang, -1.0, 1.0)
    return np.degrees(np.arccos(cosang))
```

```python
from google.colab import files

uploaded = files.upload()
df = pd.read_csv('datos_cuello.csv')
df.head()
```

Las coordenadas $(x, y)$ extraídas de las imágenes se procesaron utilizando el siguiente script para iterar sobre las filas, calcular el ángulo de cada postura y exportar las métricas descriptivas y los ROMs correspondientes.

```python
angulos = []
df.columns = df.columns.str.strip()

# Bucle para procesar cada fila del CSV
for index, row in df.iterrows():
    # Extraer coordenadas y convertir a float, reemplazando comas por puntos
    Tragus = [float(str(row['xT']).replace(',', '.')), float(str(row['yT']).replace(',', '.'))]
    C7 = [float(str(row['xC']).replace(',', '.')), float(str(row['yC']).replace(',', '.'))]
    Acromion = [float(str(row['xA']).replace(',', '.')), float(str(row['yA']).replace(',', '.'))]

    # Calcular vectores (C7 -> Tragus y C7 -> Acromion)
    v1 = np.array(Tragus) - np.array(C7)
    v2 = np.array(Acromion) - np.array(C7)

    # Usar la función para calcular el ángulo
    theta = angle_deg(v1, v2)
    angulos.append(theta)

# Añadir los ángulos calculados al DataFrame original
df['theta'] = angulos

# Exportar resultados por muestra a un nuevo CSV
df.to_csv('resultados_por_muestra.csv', index=False)
print("Archivo 'resultados_por_muestra.csv' guardado correctamente.")

```

### 4.1. Procesamiento de Coordenadas y Resultados Brutos
En la siguiente tabla se presentan las coordenadas bidimensionales extraídas de Kinovea para los marcadores anatómicos (Tragus, C7 y Acromion), correspondientes a las tres repeticiones registradas para cada una de las posturas (Neutra, Flexión y Extensión). En la última columna se muestra el ángulo cervical absoluto ($\theta$) calculado matemáticamente para cada fotograma mediante el script de procesamiento.

### 4.2. Análisis Estadístico y Dispersión
Para evaluar la consistencia de las mediciones, se agruparon los datos por postura y se calcularon las medidas de tendencia central y dispersión.Como se observa en la tabla resumen, la posición neutra estableció un ángulo base promedio de 143.27°. Los coeficientes de variación (CV) se mantuvieron bajos en todas las posturas (inferiores al 4%), indicando una alta repetibilidad del método. La siguiente gráfica ilustra visualmente estos promedios junto con su respectiva desviación estándar.

```python
# Calcular Resumen (Media, SD)
resumen = df.groupby('postura')['theta'].agg(
    media_theta='mean',
    sd_theta='std'
).reset_index()

# Calcular Coeficiente de Variación (CV = (Desviación Estándar / Media) * 100)
resumen['cv_theta'] = (resumen['sd_theta'] / resumen['media_theta']) * 100

# Exportar el resumen a un nuevo CSV
resumen.to_csv('resumen_por_postura.csv', index=False)
print("Archivo 'resumen_por_postura.csv' generado correctamente.")

# Mostrar tabla de resumen
print("\n--- Resumen Estadístico por Postura ---")
print(resumen.to_string(index=False))
```

 ### 4.3. Rango de Movimiento (ROM)
A partir de los promedios calculados, se determinó el Rango de Movimiento (ROM) relativo restando el ángulo de la posición neutra de referencia a los ángulos de flexión y extensión máximas. Los resultados arrojan un ROM de flexión de -32.41°  y un ROM de extensión anómalamente reducido de -2.03°  (fenómeno abordado en la sección de discusión debido a compensaciones torácicas). La siguiente figura representa gráficamente la magnitud de estos rangos.

```python
try:
    # Extraer las medias de la tabla resumen generada en el paso anterior
    thetaN = resumen.loc[resumen['postura'] == 'Neutra', 'media_theta'].values[0]
    thetaF = resumen.loc[resumen['postura'] == 'Flexion', 'media_theta'].values[0]
    thetaE = resumen.loc[resumen['postura'] == 'Extension', 'media_theta'].values[0]

    # Calcular ROM respecto a la posición neutra
    rom_flex = thetaF - thetaN
    rom_ext = thetaE - thetaN

    # Mostrar resultados finales
    print("--- Resultados de Rango de Movimiento (ROM) ---")
    print(f"Ángulo Medio Neutro: {thetaN:.2f} grados")
    print(f"ROM Flexión:         {rom_flex:.2f} grados")
    print(f"ROM Extensión:       {rom_ext:.2f} grados")
    print("-----------------------------------------------")

except IndexError:
    print("Advertencia: No se pudieron calcular todos los ROMs.")
    print("Asegúrese de que las posturas 'Neutra', 'Flexion' y 'Extension' estén presentes en los datos.")
```
