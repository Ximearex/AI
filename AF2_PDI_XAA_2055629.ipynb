## DESARROLLO.

### Parte teórica.

#### 1. Definiciones clave.

- **Preprocesamiento de imágenes digitales:** El preprocesamiento de imágenes es un conjunto de técnicas que se emplean para mejorar la calidad y extraer información relevante de las imágenes digitales antes de que sean analizadas y procesadas por algoritmos de visión artificial o de aprendizaje automático. Implica una serie de operaciones destinadas a corregir, filtrar, normalizar o mejorar las imágenes con el fin de optimizarlas para su posterior análisis.

- **Segmentación de imágenes digitales:** es una técnica de procesamiento que consiste en dividir la imagen en regiones o segmentos de interés para facilitar su observación y estudio. La información de cada región se obtiene a partir de criterios de similitud como la intensidad o la textura. Es una de las etapas más importantes, pues es la que permite analizar por separado los objetos de interés de la imagen.

- **Análisis de características geométricas:** estas características concentran la información paramétrica de una región de interés. Estudiarlas permite describir la forma y el tamaño de los elementos de la imagen (por ejemplo su área o su perímetro, medidos en píxeles). Cuando este análisis se aplica a una imagen ya segmentada, es posible comparar cómo cambian esas medidas en el objeto de interés respecto a la imagen original.

- **Diferencias entre los métodos de procesamiento en Fiji y OpenCV:**

**FIJI**

Fiji es una herramienta de procesamiento y análisis de imágenes científicas basada en ImageJ, diseñada principalmente para investigadores. Su principal ventaja es que cuenta con una interfaz gráfica intuitiva y una amplia colección de complementos que permiten realizar diferentes tipos de análisis sin necesidad de tener conocimientos avanzados de programación.

Se utiliza principalmente en áreas como la microscopía, histología y biología celular, donde es necesario estudiar, medir y cuantificar estructuras presentes en imágenes. Entre sus capacidades destacan la segmentación de imágenes, medición de áreas, identificación de estructuras, análisis de fluorescencia y procesamiento de imágenes multidimensionales.

**OpenCV**

OpenCV es una biblioteca de código abierto especializada en visión artificial y procesamiento de imágenes, cuyo objetivo principal es proporcionar herramientas para desarrollar aplicaciones capaces de interpretar y analizar información visual. A diferencia de Fiji, no es un programa con una interfaz gráfica de análisis, sino una biblioteca que se utiliza mediante programación, principalmente con lenguajes como Python y C++.

Su principal fortaleza es la posibilidad de implementar algoritmos personalizados y automatizar tareas de reconocimiento y detección de objetos, seguimiento de movimiento, segmentación y procesamiento de vídeo en tiempo real. Se utiliza en campos como la robótica, automatización industrial y desarrollo de tecnologías médicas.

**Importancia del análisis de imágenes en la visión artificial y la automatización.**

El análisis de imágenes es fundamental en la visión artificial porque hace posible que un sistema reconozca e interprete imágenes por sí solo y convierta esa información visual en decisiones automáticas. Al evaluar cada píxel de la imagen capturada, el sistema puede detectar patrones que se repiten, medir dimensiones y clasificar objetos con gran exactitud y rapidez, lo que reduce costos y disminuye los errores humanos.

##### 1. Preprocesamiento

Para la realización de esta fase de la actividad volveremos a utilizar la imagen del avión con fondo homogéneo proporcionada por el profesor.

Utilizaremos diversas librerías que nos permitirán realizar nuestras operaciones, pero la principal es **OpenCV**, una librería que permite procesar imágenes y extraer características usando Python.

También utilizaremos dos librerías extra: **NumPy** es una librería utilizada para manejar la imagen como una matriz numérica (2D o 3D), ya que cada imagen se representa como un arreglo de píxeles; esto porque OpenCV internamente trabaja con arreglos de NumPy. Finalmente, **Matplotlib** se emplea para visualizar la imagen en pantalla de manera clara.

```python
# ANALISIS DE UNA IMAGEN
# LIBRERIAS
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

Después cargamos la imagen con el comando `cv2.imread` y la abrimos con la librería Matplotlib mediante `plt.show` para verla en color y en escala de grises. Hay que considerar que OpenCV lee la imagen en formato **BGR (Blue, Green, Red)**, pero Matplotlib la muestra como **RGB**, por lo que la imagen no convertida se ve más azulada; esta parte del código solo es para mostrar ambas imágenes antes de todo el procedimiento.

```python
# CARGAR IMAGEN A COLOR Y A ESCALA DE GRISES
imagencolor = cv2.imread("avion.jpg")
imagengris = cv2.imread("avion.jpg", cv2.IMREAD_GRAYSCALE)
    # ABRIR IMAGEN A COLOR Y A ESCALA DE GRISES
plt.subplot(1,2,1)
plt.imshow(imagencolor)
plt.title("Imagen a Color (BGR)")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(imagengris, cmap="gray")
plt.title("Imagen en Escala de Grises")
plt.axis("off")

plt.show()
```

Antes de una modificación a la imagen buscaremos obtener información de esta misma, en este caso respecto a sus dimensiones (tamaño de la imagen en píxeles).

```python
dim_img_avion = imagencolor.shape  # alto x ancho
print(f'Dimensiones de la imagen: {dim_img_avion}')
```

Para el preprocesamiento de la imagen, convertiremos la imagen a color (BGR por OpenCV) a escala de grises para reducir información innecesaria de la imagen, además de facilitar la aplicación de la segmentación próximamente. Se realizará mediante el comando `cv2.cvtColor(imagencolor, cv2.COLOR_BGR2GRAY)`.

```python
# CONVERTIR A ESCALA DE GRISES
imagen_escdegrises = cv2.cvtColor(imagencolor, cv2.COLOR_BGR2GRAY)
print(f'Dimensiones de la imagen: {imagen_escdegrises.shape}')
```

Después de convertir a escala de grises la imagen, modificaremos y ajustaremos los valores de brillo y contraste de la imagen para permitirnos resaltar el objeto de interés. Para este paso podemos definir valores entre 0 - 255 como mínimos y máximos con `cv2.NORM_MINMAX` para poder simular el ajuste manual que realizamos en el software Fiji (100, 203).

```python
# AJUSTE DE BRILLO Y CONTRASTE
min_val = 100
max_val = 203
img_clip = np.clip(imagen_escdegrises, min_val, max_val)
img_norm = cv2.normalize(img_clip, None, 0, 255, cv2.NORM_MINMAX)

plt.imshow(img_norm, cmap="gray")
plt.title("Ajuste tipo Minimum/Maximum")
plt.axis("off")
plt.show()
```

##### 2. Segmentación

La segmentación de la imagen nos permitirá separar el objeto de interés (el avión) del fondo. En este caso, se realizará una segmentación mediante el método de umbralización (Threshold). Se aplicará por medio del comando `cv2.threshold(img_norm, valor_umbral, 255, cv2.THRESH_BINARY_INV)`, en donde el `valor_umbral` será el utilizado también en el software Fiji: 254.

```python
# SEGMENTACION
    # THRESHOLD
valor_umbral = 254
_, img_segmentada = cv2.threshold(img_norm, valor_umbral, 255, cv2.THRESH_BINARY_INV)

plt.imshow(img_segmentada, cmap='gray')
plt.title("Imagen Segmentada")
plt.axis("off")
plt.show()
```

En la imagen a color, podemos separar los canales RGB de la imagen con el comando `b, g, r = cv2.split(imagencolor)`.

```python
# SEPARAR CANALES RGB
b, g, r = cv2.split(imagencolor)

plt.subplot(1,3,1)
plt.imshow(b, cmap='gray')
plt.title('Canal azul')

plt.subplot(1,3,2)
plt.imshow(g, cmap='gray')
plt.title('Canal verde')

plt.subplot(1,3,3)
plt.imshow(r, cmap='gray')
plt.title('Canal rojo')

plt.show()
```

##### 3. Análisis de la imagen

Debido a que el avión es detectado ahora como una región separada y aislada, se puede realizar un **análisis** de esta como una nueva imagen, pudiendo medir sus propiedades.

Podemos obtener nueva información como el área del objeto segmentado. Este análisis se realiza mediante el comando `np.count_nonzero(img_segmentada)`.

```python
# AREA DEL OBJETO SEGMENTADO
total_pix_img_seg = img_segmentada.size
pix_objeto = np.count_nonzero(img_segmentada)

print(f'Área del objeto: {pix_objeto} píxeles')
print(f'Porcentaje de imagen ocupada: {pix_objeto/total_pix_img_seg*100:.3f}%')
```

Podemos también conocer sus tonos de grises, esto mediante la obtención del máximo y mínimo del valor en la escala de grises utilizado por la imagen segmentada.

```python
img_gray_min = img_segmentada.min()  # Valor mínimo de gris en la imagen
img_gray_max = img_segmentada.max()  # Valor máximo de gris en la imagen
print(f'Valor mínimo de gris: {img_gray_min}')
print(f'Valor máximo de gris: {img_gray_max}')
```

Una vez segmentado el objeto de interés mediante umbralización, se procedió a calcular su **Bounding Box**. Esta función analiza la región blanca de la imagen segmentada y calcula el rectángulo más pequeño posible que encierra completamente todos los píxeles pertenecientes al objeto segmentado. Nos devuelve parámetros de coordenadas (x, y) y dimensiones del rectángulo. Se realiza por medio del comando `x, y, w, h = cv2.boundingRect(img_segmentada)`.

```python
# BOUNDING BOX
    # EXTRAER
x, y, w, h = cv2.boundingRect(img_segmentada)
print(f"Bounding Box x: {x}, y: {y}, ancho: {w}, alto: {h}")
```

Una vez obtenidas las coordenadas del Bounding Box, este se dibuja sobre la imagen original a color; esta acción nos permite visualizar la localización espacial del objeto dentro de la imagen. Se realiza mediante el comando `cv2.rectangle(imagencolor, (x, y), (x + w, y + h), (255, 0, 0), 2)`.

```python
# DIBUJAR
img_con_boundingbox = cv2.rectangle(imagencolor, (x, y), (x + w, y + h), (255, 0, 0), 2)

plt.imshow(img_con_boundingbox, cmap='gray')
plt.title('Bounding Box del objeto segmentado')
plt.show()
```
